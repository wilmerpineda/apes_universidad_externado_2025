## ¿Qué es el análisis de regresión?

> **Modelo general de regresión:**
>  
> $$
> Y = f(X_0, X_1, \ldots, X_p; \beta_0, \beta_1, \ldots, \beta_p) + \varepsilon
> $$

El **análisis de regresión** es una colección de técnicas estadísticas cuyo propósito es estudiar y modelar la relación entre una variable de interés, denominada **respuesta** (también conocida como *variable dependiente* o *output*), y un conjunto de una o más variables llamadas **predictoras** (también conocidas como *variables independientes*, *explicativas* o *inputs*).

Estas herramientas permiten identificar patrones, cuantificar relaciones, y utilizar el conocimiento derivado del modelo para diversos fines dentro del análisis de datos.

---

### Posibles objetivos de un análisis de regresión

A través del análisis de regresión, es posible alcanzar diversos objetivos, entre los cuales destacan:

1. **Predicción**  
   Estimar el valor de la variable respuesta para nuevas observaciones de las variables explicativas.

2. **Evaluación de relaciones**  
   Determinar la magnitud, dirección e importancia del efecto que tienen las variables explicativas sobre la respuesta.

3. **Comprensión de la estructura de los datos**  
   Obtener una visión general del comportamiento de las variables en conjunto, detectando posibles patrones, tendencias o anomalías.


## 📊 Definición Matemática del Modelo Lineal

Un **modelo lineal** con $p$ variables predictoras se define como:

$$
Y_i = \beta_0 + \beta_1 X_{i1} + \beta_2 X_{i2} + \dots + \beta_p X_{ip} + \varepsilon_i
$$

En notación matricial:

$$
\mathbf{Y} = \mathbf{X} \boldsymbol{\beta} + \boldsymbol{\varepsilon}
$$

donde:

- $\mathbf{Y} \in \mathbb{R}^{n \times 1}$ es el vector de respuestas.
- $\mathbf{X} \in \mathbb{R}^{n \times (p+1)}$ es la **matriz de diseño** con una columna de unos para el intercepto.
- $\boldsymbol{\beta} \in \mathbb{R}^{(p+1) \times 1}$ es el vector de coeficientes.
- $\boldsymbol{\varepsilon} \in \mathbb{R}^{n \times 1}$ representa los errores aleatorios.

### 🔎 ¿Qué buscamos en el modelo?
- **Estimaciones de los coeficientes** $\boldsymbol{\beta}$ que minimicen los errores.
- **Validación de los supuestos** que garanticen la validez del modelo.

 La estimación de los parámetros $\beta$ puede caracterizarse de manera general como,
 
$$
 \hat{\mathbf{\beta}}=\arg\min S(\boldsymbol{\beta}),
$$

donde

$$
    S(\boldsymbol{\beta}) = \sum\limits_{i=1}^{n} M(\varepsilon_i)
$$

Pasemos a los **supuestos fundamentales** del modelo lineal.

## 🔍 Criterio de Estimación: Mínimos Cuadrados Ordinarios

 Tomando $M(h)=h^2$, la estrategia de estimación se llama  *mínimos cuadrados ordinarios*, donde el objetivo es **encontrar los coeficientes $\boldsymbol{\beta}$ que minimicen la suma de los cuadrados de los errores**:

$$
S(\boldsymbol{\beta}) = \sum_{i=1}^{n} \varepsilon_i^2 = (\mathbf{Y} - \mathbf{X} \boldsymbol{\beta})^T (\mathbf{Y} - \mathbf{X} \boldsymbol{\beta})
$$

Para minimizar esta función, derivamos con respecto a $\boldsymbol{\beta}$ y resolvemos:

$$
\frac{\partial S}{\partial \boldsymbol{\beta}} = -2\mathbf{X}^T (\mathbf{Y} - \mathbf{X} \boldsymbol{\beta}) = 0
$$

Resolviendo para $\boldsymbol{\beta}$:

$$
\mathbf{X}^T \mathbf{X} \boldsymbol{\beta} = \mathbf{X}^T \mathbf{Y}
$$

Si la matriz $\mathbf{X}^T \mathbf{X}$ es **invertible**, podemos despejar:

$$
\boldsymbol{\beta} = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{Y}
$$

Esta es la **solución de mínimos cuadrados ordinarios (OLS)**.

---

## 🧪 Propiedades básicas de inferencia

- Si $\hat{\beta}$ es el estimador por **mínimos cuadrados** de $\beta$, entonces es un **estimador insesgado**.
- Si suponemos que el modelo con variable de respuesta continua, descrito en la forma anterior, tiene **errores no sesgados, no correlacionados y con varianza constante**, entonces se cumple el siguiente resultado:

> ### Teorema de Gauss-Markov
> El estimador lineal insesgado de mínima varianza de $\beta$ es:
>
> $$
> \hat{\beta} = (\mathbf{X}^{\prime}\mathbf{X})^{-1}\mathbf{X}^{\prime}\mathbf{Y}
> $$
>
> Es decir, el estimador por mínimos cuadrados.

---

## Propiedades adicionales bajo normalidad

- Si, además, suponemos que los errores siguen una **distribución normal**, es decir, $\varepsilon \sim N(\mathbf{0}, \sigma^2 \mathbf{I})$, entonces $\hat{\beta}$ coincide con el **estimador de máxima verosimilitud**, notado $\hat{\beta}_{MV}$.
- En ausencia de normalidad, el estimador por mínimos cuadrados **no es eficiente**, es decir, tiene una varianza superior a la del estimador de máxima verosimilitud.
- En el modelo actual, un estimador **insesgado** de $\sigma^2$, notado $s^2$, está dado por:

> $$
> s^2 = \frac{\mathbf{e}^{\prime} \mathbf{e}}{n - (p+1)} = \frac{(\mathbf{Y} - \mathbf{X} \beta)^{\prime} (\mathbf{Y} - \mathbf{X} \beta)}{n - (p+1)}
> $$

- $\hat{\beta}$ y $s^2$ son **estadísticamente independientes** (según el **Teorema de Basu**).
- La siguiente variable aleatoria sigue una **distribución chi-cuadrado**:

> $$
> \frac{(n - (p+1))s^2}{\sigma^2} \sim \chi^2_{n - (p+1)}
> $$
